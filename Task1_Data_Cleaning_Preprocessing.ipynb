# Task 1: Data Cleaning & Preprocessing

```python
import pandas as pd
import numpy as np
```

## 1. Import the dataset

```python
data = {'Name': ['Aman', 'Sara', 'Ravi', 'Neha', 'Aman', 'Imran', 'Priya', 'Kabir'],
'Age': ['21', '22', None, '20', '21', '23', '24', '19'],
'City': ['lucknow', 'Delhi', 'DELHI', 'Lucknow ', 'lucknow', 'mumbai', 'Mumbai', 'delhi '],
'Marks': [85, 91, 78, None , 85, 88, 94, 72],
'Email': ['aman@example.com', 'sara@example.com', 'ravi@example.com', 'neha@example.com','aman@example.com', None, 'priya@example.com', 'kabir@example.com']
}
df = pd.DataFrame(data)
df.to_csv('raw_dataset.csv', index=False)  # Save the example raw data
df
```

## 2. Inspect the dataset structure

```python
print('First 5 rows:')
display(df.head())
print('Shape (rows, columns):',
df.shape)
print('\nColumn names:')
print(df.columns.tolist())
print('\nData types and non-null counts:')
df.info()
print('\nStatistical summary:')
display(df.describe(include='all'))
```

## 3. Identify missing values and duplicates

```python
print('Missing values per column:')
print(df.isnull().sum())
print('\nNumber of duplicate rows:', df.duplicated().sum())
print('\nDuplicate rows:')
display(df[df.duplicated(keep=False)])
print('\nUnique values in City before cleaning:')
print(df['City'].unique())
```

## 4. Clean the dataset

```python
clean_df = df.copy()

# Standardize text values
clean_df['Name'] = clean_df['Name'].astype('string').str.strip().str.title()
clean_df['City'] = clean_df['City'].astype('string').str.strip().str.title()
clean_df['Email'] = clean_df['Email'].astype('string').str.strip().str.lower()

# Correct the known inconsistent age entry
clean_df['Age'] = clean_df['Age'].replace({'twenty-three': '23'})

# Correct data types; invalid values become NaN
clean_df['Age'] = pd.to_numeric(clean_df['Age'], errors='coerce')
clean_df['Marks'] = pd.to_numeric(clean_df['Marks'], errors='coerce')

# Fill missing numeric values with the median
clean_df['Age'] = clean_df['Age'].fillna(clean_df['Age'].median()).round().astype('int64')
clean_df['Marks'] = clean_df['Marks'].fillna(clean_df['Marks'].median())

# Remove exact duplicate rows
clean_df = clean_df.drop_duplicates().reset_index(drop=True)

display(clean_df)
```

## 5. Verify the cleaned dataset

```python
print('Shape after cleaning:', clean_df.shape)
print('\nMissing values after cleaning:')
print(clean_df.isnull().sum())
print('\nDuplicate rows after cleaning:', clean_df.duplicated().sum())
print('\nFinal data types:')
print(clean_df.dtypes)
print('\nStandardized city values:', clean_df['City'].unique())
```

## 6. Save the cleaned dataset as a new CSV file

```python
output_file = 'cleaned_dataset.csv'
clean_df.to_csv(output_file, index=False)
print(f'Cleaned dataset saved successfully as: {output_file}')
```
